```python
import numpy as np
from emu_like.datasets import Dataset
```

```python
for spectrum in ['pk_m', 'pk_cb', 'pk_weyl', 'fk_m', 'fk_cb', 'fk_weyl']:
    for data_range in ['thin', 'std', 'ext']:
        data = Dataset().load(
            path='/data/emilio/emu_like/lcdm/sample/pk_100_{}.fits'.format(data_range),
            name=spectrum,
            verbose=False,
        )
        y_min = data.y.min()
        y_max = data.y.max()
        if np.sign(y_min) == np.sign(y_max):
            sign = np.sign(y_min)
        else:
            sign = '+/-'
        print('{} - {} - {}: min {:.2e}, max {:.2e}'.format(sign, spectrum, data_range, y_min, y_max))
```
