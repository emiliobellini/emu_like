```python
import numpy as np
import os
from emu_like.datasets import Dataset

root = '/ceph/hpc/data/s25r06-05-users'
```

```python
model = 'lcdm'

print('Model {}'.format(model))

for spectrum in ['pk_m', 'pk_cb', 'pk_weyl', 'fk_m', 'fk_cb', 'fk_weyl']:
    for data_range in ['thin', 'std', 'ext']:
        data = Dataset().load(
            path=os.path.join(root, '{}/sample/pk_100_{}.fits'.format(model, data_range)),
            name=spectrum,
            verbose=False,
        )
        y_min = data.y.min()
        y_max = data.y.max()
        if np.sign(y_min) == np.sign(y_max):
            sign = np.sign(y_min)
        else:
            sign = '+/-'
        print('{} - {} - {}: min {:.2e}, max {:.2e}'.format(sign, spectrum, data_range, y_min, y_max))

print()

for spectrum in ['cl_TT_lensed', 'cl_TE_lensed', 'cl_EE_lensed', 'cl_BB_lensed', 'cl_Tp_lensed', 'cl_pp_lensed']:
    for data_range in ['thin', 'std', 'ext']:
        data = Dataset().load(
            path=os.path.join(root, '{}/sample/cl_100_{}.fits'.format(model, data_range)),
            name=spectrum,
            verbose=False,
        )
        y_min = data.y.min()
        y_max = data.y.max()
        if np.sign(y_min) == np.sign(y_max):
            sign = np.sign(y_min)
        else:
            sign = '+/-'
        print('{} - {} - {}: min {:.2e}, max {:.2e}'.format(sign, spectrum, data_range, y_min, y_max))
```
